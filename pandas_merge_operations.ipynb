# Question Practise

```python
import pandas as pd

# Customers
customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4, 5],
    "name": ["A", "B", "C", "D", "E"],
    "city": ["Bangalore", "Delhi", "Mumbai", "Chennai", "Hyderabad"]
})

# Orders
orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105, 106],
    "customer_id": [1, 2, 2, 3, 6, 2],
    "amount": [500, 300, 200, 400, 800, 150]
})

# Payments
payments = pd.DataFrame({
    "payment_id": [201, 202, 203, 204],
    "order_id": [101, 102, 104, 107],
    "status": ["Paid", "Pending", "Paid", "Failed"]
})
```

```python
# Q1. Inner Join Basics
# Merge customers and orders to show only customers who have placed orders.
# Output: customer_id, name, order_id, amount
```

```python
# @title
q1 = pd.merge(customers, orders, on="customer_id", how="inner")[["customer_id", "name", "order_id", "amount"]]
print(q1)
```

```python
# Q2. Left Join (Unmatched Handling)
# Show all customers and their orders (if any).
# Identify customers who never placed an order
```

```python
# @title
q2 = pd.merge(customers, orders, on="customer_id", how="left")
print(q2)
```

```python
# Q3. Right Join Insight
# Show all orders with customer details.
# Find orders that do not have a valid customer
```

```python
# @title
q3 = pd.merge(customers, orders, on="customer_id", how="right")
print(q3)
```

```python
# @title
invalid_orders = q3[q3["name"].isna()][["order_id", "customer_id"]]
print(invalid_orders)
```

```python
# Q4. Full Outer Join Analysis
# Perform an outer join between customers and orders.
# Count how many rows have missing values
# Explain why they occur
```

```python
# @title
q4 = pd.merge(customers, orders, on="customer_id", how="outer")
print(q4)
missing_count = q4.isna().sum().sum()
print("-----------------------")
print(missing_count)
```

```python
# @title
temp = pd.merge(customers, orders, on="customer_id", how="inner")

q5 = pd.merge(temp, payments, on="order_id", how="left")[
    ["name", "order_id", "amount", "status"]
]
print(q5)
```

```python
# Q6. Cross Join Challenge 🔥
# Create all possible combinations of:
# customers × payment status
# Questions:
# Total number of rows?
# Why is this useful?
```

```python
# @title
q6 = customers.merge(payments, how="cross")
print(q6)
```

```python
# Q7. Overlapping Columns + Cleanup
# Create a new DataFrame:
# customer_info = customers.rename(columns={"city": "location"})
# Now:
# Merge customers and customer_info
# Handle duplicate columns properly
# Clean final output (no _x, _y)
```

```python
# @title
customer_info = customers.rename(columns={"city": "location"})

q7 = pd.merge(customers, customer_info, on="customer_id")
print(q7)
print("--------------------------------------")

# Clean duplicate columns
q7 = q7.drop(columns=["name_y"]).rename(columns={"name_x": "name"})

print(q7)
```
